```python
# first just grab the file. Using the code from the M1 HW here.

# Cleaning up the HTML and such.

import requests, re, nltk
from bs4 import BeautifulSoup
from nltk import clean_html
from collections import Counter
import operator

# nltk.download('stopwords')#uncomment if not yet downloaded

# we may not care about the usage of stop words
stop_words = nltk.corpus.stopwords.words('english') + [
 'ut', '\'re','.', ',', '--', '\'s', '?', ')', '(', ':', '\'', '–', '—', ''
 '\"', '-', '}', '{', '&', '|', u'\u2014' ]

# We most likely would like to remove html markup
def cleanHtml (html):
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(html, 'html.parser')
    return soup .get_text()

# We also want to remove special characters, quotes, etc. from each word
def cleanWord (w):
    # r in r'[.,"\']' tells to treat \ as a regular character 
    # but we need to escape ' with \'
    # any character between the brackets [] is to be removed 
    wn = re.sub('[,"\.\'&\|:@>*;/=”’“‘?()\]_[!]', "", w)
    # get rid of numbers at the end of a string, e.g. "word1" -> "word"
    wn = re.sub('[0-9]+$', "", wn)
    # get rid of numbers
    return re.sub('^[0-9\.]*$', "", wn)
       
# define a function to get text/clean/calculate frequency
def get_wds (URL):
    # first get the web page
    r = requests .get(URL)
    
    # Now clean
    # remove html markup
    t = cleanHtml (r .text) .lower()
    
    # split string into an array of words using any sequence of spaces "\s+" 
    wds = re.split('\s+',t)
    #split on dashes too
    wds = re.split('[-—\s]+',t)
    
    # remove periods, commas, etc stuck to the edges of words
    for i in range(len(wds)):
        wds [i] = cleanWord (wds [i])

    return wds
    
# just calculate freq
def calc_freq(wds):
    # now count the words
    wf = Counter(wds)
    
    # remove stop words
    for w in stop_words:
        if w in wf:
            del wf[w]
    
    # sort by frequency
    wf = sorted(wf.items(), key=operator.itemgetter(1), reverse=True)
    
    # get total number of words
    tw = sum(wf[i][1] for i in range(len(wf)))
    
    return (wf, tw)
        
# Grab all the words
wds = get_wds('https://www.gutenberg.org/cache/epub/1727/pg1727.txt')

# specific to our book, removing the introduction.
wds = wds[2967:]

# and now removing the footnotes.
wds = wds[:wds.index('footnotes')]

print("Total words in the Odyssey: ", len(wds))
```

```python
# breaking down by book.

books = re.split('book', ' '.join(wds))

# need to pop the first book, it is an empty list.
books.pop(0)

print("Total number of books: ", len(books)) # we are expecting 24.

# sanity check, should say i the gods in council
print(books[0][:22])
```

```python
# for an example. wf for each book.
```

```python
non_alpha_words = set()

for word in wds:
    if word.isalpha() == False:
        non_alpha_words.add(word)

non_alpha_words
```

```python
# remove the stop words

no_stop_wds = [w for w in wds if w not in stop_words]

len(no_stop_wds)
```

```python
(wf, tw) = calc_freq(no_stop_wds)

print("Total words in the Odyssey: ", tw)   # this has stop words removed btw.
print("Top words in the Odyssey: ", wf[:15])
```

```python
# breaking down by book then removing stop words and calculating frequency.

book_freqs = []
for book in books:
    no_stop_wds = [w for w in book.split() if w not in stop_words]
    (wf, tw) = calc_freq(no_stop_wds)
    book_freqs.append((wf, tw))

print("Total words in each book: ", [tw for (wf, tw) in book_freqs])
```
